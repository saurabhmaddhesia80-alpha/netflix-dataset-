# **Top 5 Countries with Most Contentt**

1. fillna('Unknown'):
Replaces missing values (NaN) in the 'country' column with the string 'Unknown'.
2. .apply(lambda x: x.split(',')[0]):
Some rows might have multiple countries separated by commas (e.g., "United States, Canada"). This takes only the first country before the comma.
3. Sets the size of the plot to be wider and taller (12 inches by 6 inches).
4. Counts how many times each country appears in the dataset and selects the top 10 countries with the most titles.
5. Creates a horizontal bar plot where:
  
      x-axis: Number of titles (counts)

      y-axis: Country names

      Uses the 'viridis' color palette for styling

6. Adds a title and axis labels, then displays the plot.

```python
Netflix['country'] = Netflix['country'].fillna('Unknown').apply(lambda x: x.split(',')[0])

plt.figure(figsize=(12,6))
top_countries = Netflix['country'].value_counts().head(5)
sns.barplot(x=top_countries.values, y=top_countries.index, palette='viridis')
plt.title('Top 5 Countries with Most Content')
plt.xlabel('Number of Titles')
plt.ylabel('Country')
plt.show()
```

# **TOP 10 CENRES**

1. Replaces any missing (NaN) values in the 'listed_in' column with the word 'Unknown'.This ensures no errors when splitting genres later.

2. str.split(','): Splits each genre string into a list, because some entries have multiple genres like:

      expand=True: Turns the split lists into separate columns temporarily.

      .stack(): Converts the wide format back to a long format (all genres in a single column).

      .value_counts(): Counts how often each genre appears.

      .head(10): Keeps only the top 10 most frequent genres.

3. Creates a wide plot area (12x6 inches), giving enough room for genre names and bars.

4. x-axis: The count of each genre (number of times it appears). y-axis: The genre names. palette='plasma': Uses a vibrant color palette for visual appeal.

5. Adds a clear title and axis labels to explain the chart.

6. Displays the final bar chart.

```python
Netflix['listed_in'] = Netflix['listed_in'].fillna('Unknown')
genres = Netflix['listed_in'].str.split(',', expand=True).stack().value_counts().head(10)
plt.figure(figsize=(12,6))
sns.barplot(x=genres.values, y=genres.index, palette='plasma')
plt.title('Top 10 Genres')
plt.xlabel('Count')
plt.ylabel('Genre')
plt.show()
```

## **Pie-chart for the Type: Movie and TV Shows**

1. Defines the labels for the pie chart slices (two categories).

2. Counts how many titles are of each type (Movie and TV Show).

3. Chooses 2 colors from the Wistia colormap to color the slices. np.linspace(0,1,2) creates 2 evenly spaced values between 0 and 1, used to pick colors.

4. Makes the second slice (“TV show”) pop out slightly for emphasis. 0 means no explosion for the first slice, 0.1 pushes the second slice out.

5. Sets the figure size to 5 inches by 5 inches (square shape for pie).

6. size: slice sizes

     labels: slice labels

     colors: slice colors

     explode: slice offsets

     shadow=True: adds shadow for 3D effect

     startangle=90: rotates the start so the first slice starts at 90 degrees (top)

7. Adds a big, clear title above the pie chart.

8. Adds a legend explaining the slices.

9. Displays the pie chart.





```python
labels = ['Movie', 'TV show']
size = Netflix['type'].value_counts()
colors = plt.cm.Wistia(np.linspace(0, 1, 2))
explode = [0, 0.1]
plt.rcParams['figure.figsize'] = (5, 5)
plt.pie(size,labels=labels, colors = colors, explode = explode, shadow = True, startangle = 90)
plt.title('Distribution of Type', fontsize = 25)
plt.legend()
plt.show()
```

# **Distribution of Movie Durations**

1. movies: Contains only rows where type is 'Movie'

      tv_shows: Contains only rows where type is 'TV Show'

2. Removes the ' min' part using str.replace()

      Converts the cleaned string to a float so we can perform numeric analysis (e.g., plotting)

3. histplot(): Creates a histogram to show distribution.

      movies['duration'].dropna(): Removes missing durations before plotting.

      bins=30: Splits the range of durations into 30 intervals.

      kde=True: Adds a smooth curve (Kernel Density Estimate) to show the distribution trend.
      
      color='skyblue': Makes the bars light blue for visual appeal.

4. Adds descriptive title and axis labels to make the chart easy to understand.

5. Displays the final histogram.

```python
movies = Netflix[Netflix['type'] == 'Movie']
tv_shows = Netflix[Netflix['type'] == 'TV Show']

plt.figure(figsize=(10,6))
movies['duration'] = movies['duration'].str.replace(' min', '').astype(float)

sns.histplot(movies['duration'].dropna(), bins=30, kde=True, color='skyblue')
plt.title('Distribution of Movie Durations')
plt.xlabel('Duration (minutes)')
plt.ylabel('Count')
plt.show()
```

# **Top 10 Directors on Netflix**

1. Netflix['director']: This column contains the names of directors.

      .dropna(): Removes rows where the director name is missing (NaN).

      .str.split(',', expand=True): Some entries have multiple directors, separated by commas. This splits those entries into separate columns.

      .stack(): Converts the wide format (multiple columns) into a single vertical column with all individual director names.

      .value_counts(): Counts how many times each director appears.

      .head(10): Gets the top 10 most frequent directors.
2. Sets the plot size to 12 inches wide and 6 inches high.

3. x-axis: Number of titles directed. y-axis: Director names,        palette='coolwarm': Uses a red-to-blue color scheme for visual appeal.

4. Title: Clearly explains what the chart shows.

      x-axis: Number of shows/movies directed.

      y-axis: Names of directors.

5. Renders the final bar chart.

```python
top_directors = Netflix['director'].dropna().str.split(',', expand=True).stack().value_counts().head(10)

plt.figure(figsize=(12,6))
sns.barplot(x=top_directors.values, y=top_directors.index, palette='coolwarm')
plt.title('Top 10 Directors on Netflix')
plt.xlabel('Number of Titles')
plt.ylabel('Director')
plt.show()
```

## **Country**

1. plt.subplots(figsize=(25,15))

* Creates a figure and axes object for the plot.

* Sets the size to 25 inches wide and 15 inches tall, so the word cloud is large and clear.
